```
import pandas as pd
import numpy as np
from scipy import stats
from joblib import Parallel, delayed
import matplotlib.pyplot as plt
```

```
df = pd.read_excel('data/dataset_set_A_aguas_residuales.xlsx')
df['fecha_registro'] = pd.to_datetime(df['fecha_registro'])
df.head()
```

```
calidad = {'filas':len(df), 'columnas':len(df.columns), 'faltantes':int(df.isna().sum().sum()), 'duplicados':int(df.duplicated().sum())}
calidad
```

```
df['eficiencia_DBO_pct']=(1-df['DBO_salida_mg_L']/df['DBO_entrada_mg_L'])*100
resumen=df.groupby('planta').agg(registros=('planta','size'),dbo_salida_promedio=('DBO_salida_mg_L','mean'),cumplimiento_pct=('cumplimiento_norma','mean'),eficiencia_dbo=('eficiencia_DBO_pct','mean')).reset_index()
resumen['cumplimiento_pct']*=100
resumen
```

```
r = df['caudal_entrada_m3_d'].corr(df['energia_aeracion_kWh'])
r
```

```
grupos=[g['DBO_salida_mg_L'].values for _,g in df.groupby('planta')]
stats.f_oneway(*grupos)
```

```
comp=df.groupby('planta')['cumplimiento_norma'].mean()*100
comp.plot(kind='bar', title='Cumplimiento por planta', ylabel='%')
plt.tight_layout()
```
